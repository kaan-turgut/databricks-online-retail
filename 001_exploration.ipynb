## Read the File

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *

schema = StructType([
    StructField("Invoice", StringType(), True),
    StructField("StockCode", StringType(), True),
    StructField("Description", StringType(), True),
    StructField("Quantity", IntegerType(), True),
    StructField("InvoiceDate", TimestampType(), True),
    StructField("Price", DecimalType(12, 4), True),
    StructField("Customer ID", DoubleType(), True),
    StructField("Country", StringType(), True)
])

df_2009_2010 = (
    spark.read
    .option("header", "true")
    .option("mode", "FAILFAST")
    .schema(schema)
    .csv("/Volumes/retail_project/landing/raw_files/Year_2009-2010.csv")
)

# The Customer IDs appear as decimals in the csv file. We need to cast them to integers
df_2009_2010 = df_2009_2010.withColumn("Customer ID", col("Customer ID").cast("int"))
```

## Exploratory Analysis

```python
df_2009_2010.show()
df_2009_2010.printSchema()
print(f"Number of rows: {df_2009_2010.count()}")
```

```python
from pyspark.sql.functions import *

df_2009_2010.select([
    count(when(col(c).isNull(), 1)).alias(c) for c in df_2009_2010.columns
]).show()
```
